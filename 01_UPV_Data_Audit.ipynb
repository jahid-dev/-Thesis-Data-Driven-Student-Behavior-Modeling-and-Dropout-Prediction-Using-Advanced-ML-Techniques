# SECTION 9: Temporal Window Analysis

```python
#Extract time windows
import re

def extract_time_windows(df):
    pattern = re.compile(r'_(\d{4})_(\d+)$')

    windows = set()

    for col in df.columns:
        match = pattern.search(col)
        if match:
            year = int(match.group(1))
            month = int(match.group(2))
            windows.add((year, month))

    return sorted(windows)


for year, df in [
    (2018, df_2018),
    (2021, df_2021),
    (2022, df_2022)
]:
    windows = extract_time_windows(df)

    print(f"\n===== {year} =====")
    print("Number of time windows:", len(windows))
    print("Time windows:")
    print(windows)
```

```python
for year, df in [
    (2018, df_2018),
    (2021, df_2021),
    (2022, df_2022)
]:

    windows = extract_time_windows(df)

    print(f"\n{year}:")
    print(" → ".join(f"{y}-{m:02d}" for y, m in windows))
```

```python

# SECTION 10A: FEATURE GROUP SUMMARY


# Feature groups are defined based on the observed UPV column
# names and temporal feature prefixes.

FEATURE_GROUP_RULES = {

    "Student Identity": [
        "dni_hash",
        "tit_hash",
        "asi_hash"
    ],

    "Admission Information": [
        "anyo_ingreso",
        "tipo_ingreso",
        "preferencia_seleccion",
        "campus_hash",
        "desplazado_hash"
    ],

    "Academic Background": [
        "estudios_p_hash",
        "estudios_m_hash",
        "dedicacion",
        "anyo_inicio_estudios",
        "es_retitulado",
        "es_adaptado"
    ],

    "Academic Performance": [
        "nota10_hash",
        "nota14_hash",
        "nota_asig_hash",
        "rendimiento_cuat_a",
        "rendimiento_cuat_b",
        "rendimiento_total",
        "rend_total_ultimo",
        "rend_total_penultimo",
        "rend_total_antepenultimo",
        "exento_npp"
    ],

    "Course Progress": [
        "curso_mas_bajo",
        "curso_mas_alto"
    ],

    "Credit Information": [
        "cred_mat1",
        "cred_mat2",
        "cred_mat3",
        "cred_mat4",
        "cred_mat5",
        "cred_mat6",
        "cred_sup_normal",
        "cred_sup_espec",
        "cred_sup",
        "cred_mat_normal",
        "cred_mat_movilidad",
        "cred_ptes_acta",
        "cred_mat_practicas",
        "cred_mat_sem_a",
        "cred_mat_sem_b",
        "cred_mat_anu",
        "cred_mat_total",
        "cred_sup_sem_a",
        "cred_sup_sem_b",
        "cred_sup_anu",
        "cred_sup_total",
        "cred_sup_1o",
        "cred_sup_2o",
        "cred_sup_3o",
        "cred_sup_4o",
        "cred_sup_5o",
        "cred_sup_6o",
        "cred_sup_tit",
        "cred_pend_sup_tit"
    ],

    "Enrollment / Administrative": [
        "matricula_activa",
        "baja_fecha",
        "impagado_curso_mat"
    ],

    "Student Activities": [
        "practicas",
        "actividades",
        "ajuste",
        "asig1",
        "pract1",
        "activ1",
        "total1",
        "ajuste1"
    ],

    "Target": [
        "abandono_hash"
    ]
}


# Temporal feature prefixes
TEMPORAL_FEATURE_PREFIXES = [
    "pft_events_",
    "pft_visits_",
    "pft_days_logged_",
    "pft_assignment_submissions_",
    "pft_test_submissions_",
    "pft_total_minutes_",
    "resource_events_",
    "n_resource_days_",
    "n_wifi_days_"
]


def get_feature_group(column):

    # Check explicitly defined columns first
    for group, columns in FEATURE_GROUP_RULES.items():
        if column in columns:
            return group

    # Check temporal/digital behaviour features
    for prefix in TEMPORAL_FEATURE_PREFIXES:
        if column.startswith(prefix):
            return "Temporal / Digital Behavior"

    # Dataset timestamp
    if column == "fecha_datos":
        return "Temporal Metadata"

    # Any column not covered above
    return "Other / Uncategorized"


def create_feature_summary(df, dataset_name):

    rows = []

    for column in df.columns:

        rows.append({
            "Dataset": dataset_name,
            "Column": column,
            "Data Type": str(df[column].dtype),
            "Feature Group": get_feature_group(column),
            "Missing Count": int(df[column].isna().sum()),
            "Missing %": round(df[column].isna().mean() * 100, 2),
            "Unique Values": int(df[column].nunique(dropna=True))
        })

    return pd.DataFrame(rows)


# Create column-level feature summaries
feature_summary_2018 = create_feature_summary(
    df_2018,
    "2018"
)

feature_summary_2021 = create_feature_summary(
    df_2021,
    "2021"
)

feature_summary_2022 = create_feature_summary(
    df_2022,
    "2022"
)


# Combine all datasets
feature_summary_all = pd.concat(
    [
        feature_summary_2018,
        feature_summary_2021,
        feature_summary_2022
    ],
    ignore_index=True
)


# Display column-level summary
print("=" * 70)
print("FEATURE GROUP SUMMARY")
print("=" * 70)

display(feature_summary_all)


# Group-level summary for each dataset
print("\n" + "=" * 70)
print("FEATURE GROUP COUNTS")
print("=" * 70)

group_summary = (
    feature_summary_all
    .groupby(
        ["Dataset", "Feature Group"]
    )
    .agg(
        Number_of_Features=("Column", "count"),
        Total_Missing=("Missing Count", "sum"),
        Average_Missing_Percent=("Missing %", "mean")
    )
    .reset_index()
)

display(group_summary)


# Show uncategorized columns
print("\n" + "=" * 70)
print("UNCATEGORIZED COLUMNS")
print("=" * 70)

uncategorized = (
    feature_summary_all[
        feature_summary_all["Feature Group"] == "Other / Uncategorized"
    ][
        ["Dataset", "Column", "Data Type"]
    ]
)

display(uncategorized)
```

# SECTION 20: Data Dictionary + Feature Meaning


```python
import re
import pandas as pd

print("=" * 70)
print("DATA DICTIONARY + FEATURE MEANING")
print("=" * 70)


# ----------------------------------------------------------------------
# Feature group rules
# ----------------------------------------------------------------------

FEATURE_GROUP_RULES = {

    "Student Identity": [
        "dni_hash",
        "tit_hash",
        "asi_hash"
    ],

    "Target": [
        "abandono_hash"
    ],

    "Admission Information": [
        "anyo_ingreso",
        "tipo_ingreso"
    ],

    "Temporal Metadata": [
        "fecha_datos"
    ],

    "Course Progress": [
        "curso_mas_bajo",
        "curso_mas_alto"
    ],

    "Withdrawal Information": [
        "baja_fecha"
    ]
}


# ----------------------------------------------------------------------
# Determine feature group
# ----------------------------------------------------------------------

def get_feature_group(column):

    # Exact match
    for group, columns in FEATURE_GROUP_RULES.items():
        if column in columns:
            return group

    # Temporal / digital behaviour
    temporal_prefixes = (
        "pft_",
        "n_resource_",
        "n_wifi_",
        "resource_events_"
    )

    if column.startswith(temporal_prefixes):
        return "Temporal / Digital Behavior"

    # Academic / credit related naming patterns
    academic_keywords = [
        "nota",
        "calificacion",
        "grade",
        "exam",
        "evaluacion",
        "rendimiento"
    ]

    if any(keyword in column.lower() for keyword in academic_keywords):
        return "Academic Performance"

    credit_keywords = [
        "credito",
        "credit"
    ]

    if any(keyword in column.lower() for keyword in credit_keywords):
        return "Credit Information"

    # Enrollment / administrative
    enrollment_keywords = [
        "matricula",
        "enrollment",
        "inscripcion",
        "administr"
    ]

    if any(keyword in column.lower() for keyword in enrollment_keywords):
        return "Enrollment / Administrative"

    # Activity related
    activity_keywords = [
        "actividad",
        "activity",
        "participacion",
        "participation"
    ]

    if any(keyword in column.lower() for keyword in activity_keywords):
        return "Student Activities"

    return "Other / Uncategorized"


# ----------------------------------------------------------------------
# Identify temporal feature
# ----------------------------------------------------------------------

def is_temporal_feature(column):

    temporal_prefixes = (
        "pft_",
        "n_resource_",
        "n_wifi_",
        "resource_events_"
    )

    return column.startswith(temporal_prefixes)


# ----------------------------------------------------------------------
# Generate feature meaning
# ----------------------------------------------------------------------

def get_feature_meaning(column):

    # Known fields
    known_meanings = {

        "dni_hash":
            "Anonymized student identifier.",

        "tit_hash":
            "Anonymized identifier associated with the academic entity.",

        "asi_hash":
            "Anonymized identifier associated with the academic record.",

        "abandono_hash":
            "Binary dropout/withdrawal target variable.",

        "anyo_ingreso":
            "Year of admission.",

        "tipo_ingreso":
            "Type or category of admission.",

        "anyo_inicio_estudios":
            "Year in which the student's studies began.",

        "fecha_datos":
            "Timestamp representing the dataset observation/data extraction date.",

        "curso_mas_bajo":
            "Lowest course level recorded for the student.",

        "curso_mas_alto":
            "Highest course level recorded for the student.",

        "baja_fecha":
            "Date associated with student withdrawal/dropout."
    }

    if column in known_meanings:
        return known_meanings[column]

    # Temporal features
    temporal_match = re.search(
        r"(.+)_(\d{4})_(\d{1,2})$",
        column
    )

    if temporal_match:

        base_name = temporal_match.group(1)
        year = temporal_match.group(2)
        month = temporal_match.group(3).zfill(2)

        meaning_map = {

            "pft_events":
                "Number of platform events",

            "pft_visits":
                "Number of platform visits",

            "pft_days_logged":
                "Number of days with platform activity",

            "pft_assignment_submissions":
                "Number of assignment submissions",

            "pft_test_submissions":
                "Number of test submissions",

            "pft_total_minutes":
                "Total platform usage time in minutes",

            "n_wifi_days":
                "Number of days with Wi-Fi activity",

            "resource_events":
                "Number of resource-related events",

            "n_resource_days":
                "Number of days with resource activity"
        }

        base_meaning = meaning_map.get(
            base_name,
            "Temporal/digital behaviour measure"
        )

        return f"{base_meaning} during {year}-{month}."

    # Fallback
    return "Meaning requires verification from the original UPV documentation/codebook."


# ----------------------------------------------------------------------
# Build data dictionary
# ----------------------------------------------------------------------

data_dictionary = []

for year, df in datasets.items():

    for column in df.columns:

        missing_count = int(df[column].isna().sum())
        missing_pct = (
            missing_count / len(df) * 100
            if len(df) > 0 else 0
        )

        data_dictionary.append({

            "Dataset": year,
            "Column": column,
            "Data Type": str(df[column].dtype),
            "Feature Group": get_feature_group(column),
            "Meaning": get_feature_meaning(column),
            "Missing Count": missing_count,
            "Missing %": round(missing_pct, 2),
            "Unique Values": int(df[column].nunique(dropna=True)),
            "Temporal Feature": is_temporal_feature(column)

        })


data_dictionary_df = pd.DataFrame(data_dictionary)


# ----------------------------------------------------------------------
# Display
# ----------------------------------------------------------------------

print("\nTotal dictionary entries:",
      len(data_dictionary_df))

display(data_dictionary_df.head(20))


print("\n" + "=" * 70)
print("FEATURE GROUP DISTRIBUTION")
print("=" * 70)

display(
    data_dictionary_df
    .groupby(["Dataset", "Feature Group"])
    .size()
    .reset_index(name="Number of Features")
)


print("\n" + "=" * 70)
print("UNCATEGORIZED FEATURES")
print("=" * 70)

uncategorized = data_dictionary_df[
    data_dictionary_df["Feature Group"] == "Other / Uncategorized"
]

display(uncategorized)
```

```python
print("=" * 70)
print("RESEARCH-READY DATASET SUMMARY")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Dataset-level summary
# ----------------------------------------------------------------------

summary_rows = []

for year, df in datasets.items():

    unique_students = (
        df["dni_hash"].nunique()
        if "dni_hash" in df.columns
        else None
    )

    target_column = "abandono_hash"

    target_available = (
        target_column in df.columns
    )

    temporal_features = sum(
        is_temporal_feature(col)
        for col in df.columns
    )

    summary_rows.append({

        "Dataset": year,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Unique Students": unique_students,
        "Temporal Features": temporal_features,
        "Target Available": target_available

    })


research_summary_df = pd.DataFrame(summary_rows)


print("\nDATASET SUMMARY")
display(research_summary_df)


# ----------------------------------------------------------------------
# 2. Common columns across all datasets
# ----------------------------------------------------------------------

dataset_column_sets = {
    year: set(df.columns)
    for year, df in datasets.items()
}

common_columns = set.intersection(
    *dataset_column_sets.values()
)

print("\n" + "=" * 70)
print("COMMON COLUMNS ACROSS ALL DATASETS")
print("=" * 70)

print(f"Number of common columns: {len(common_columns)}")

print("\nCommon columns:")
for column in sorted(common_columns):
    print(column)


# ----------------------------------------------------------------------
# 3. Dataset-specific columns
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET-SPECIFIC COLUMNS")
print("=" * 70)

for year in datasets:

    other_years = [
        y for y in datasets
        if y != year
    ]

    other_columns = set.union(
        *(dataset_column_sets[y] for y in other_years)
    )

    unique_columns = (
        dataset_column_sets[year] - other_columns
    )

    print(
        f"\n{year}: "
        f"{len(unique_columns)} dataset-specific columns"
    )


# ----------------------------------------------------------------------
# 4. Target readiness
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET READINESS")
print("=" * 70)

for year, df in datasets.items():

    if "abandono_hash" not in df.columns:
        print(f"{year}: Target column NOT FOUND")
        continue

    target_counts = df["abandono_hash"].value_counts()

    print(f"\nDataset {year}")
    print("Target values:")
    print(target_counts.to_dict())

    print(
        "Target missing:",
        int(df["abandono_hash"].isna().sum())
    )


# ----------------------------------------------------------------------
# 5. Identifier handling recommendation
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("MODELING PREPARATION NOTES")
print("=" * 70)

print("""
1. dni_hash is an anonymized student identifier and should not be
   directly used as a predictive feature.

2. Target variable:
   abandono_hash

3. baja_fecha should be treated carefully because it is directly
   associated with withdrawal/dropout information and may introduce
   target leakage depending on the prediction time.

4. fecha_datos is dataset-level temporal metadata and should not
   automatically be treated as a student behavioural feature.

5. Temporal/digital behaviour variables require temporal alignment
   before combining datasets from different academic periods.

6. Features with extremely high missingness should be evaluated
   before model training.

7. Dataset-specific columns should not automatically be discarded;
   their research relevance and temporal comparability should first
   be evaluated.

8. The final modeling dataset should be created only after feature
   selection, leakage analysis, missing-value treatment, encoding,
   and temporal alignment.
""")


# ----------------------------------------------------------------------
# 6. Final audit status
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL STATUS")
print("=" * 70)

print("""
The UPV datasets are structurally suitable for further research
after appropriate preprocessing.

The audit confirms:

- Multiple longitudinal records per student
- Consistent binary target within each student
- Temporal/digital behavioural features
- Different temporal observation windows across datasets
- Dataset-specific temporal columns
- Missing values that require treatment
- Potential leakage-sensitive variables such as withdrawal date

Therefore, the next stage is feature engineering and preprocessing,
not direct model training.
""")
```
